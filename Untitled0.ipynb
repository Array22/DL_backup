```python
# Move UIEB images to TCTL data folder

import os
import shutil

source_dir = [
    "/content/drive/MyDrive/dataset/UIEB Dataset/reference/UIEB_reference_reName",
    "/content/drive/MyDrive/dataset/UIEB Dataset/raw/UIEB_raw_reName"
]

destination_dir = [
    "/content/drive/MyDrive/TCTL/data/ref",
    "/content/drive/MyDrive/TCTL/data/raw"
]

for src, dst in zip(source_dir, destination_dir):

    if not os.path.isdir(src):
        print(f"Error: Source directory does not exist: {src}")
        continue

    os.makedirs(dst, exist_ok=True)

    for item in os.listdir(src):
        src_path = os.path.join(src, item)
        dst_path = os.path.join(dst, item)

        if os.path.isdir(src_path):
            shutil.copytree(src_path, dst_path, dirs_exist_ok=True)
        else:
            shutil.copy2(src_path, dst_path)

    print(f"Copied: {src}")
    print(f"     → {dst}")
```

```python
# Generate necessary txt files (color transfer parameters)

import os
import cv2
import numpy as np

ref_dir = "/content/drive/MyDrive/TCTL/data/ref"
txt_dir = "/content/drive/MyDrive/TCTL/data/txt"

output_size = (256, 256)

os.makedirs(txt_dir, exist_ok=True)

image_extensions = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff"}

for filename in sorted(os.listdir(ref_dir)):

    name, ext = os.path.splitext(filename)

    if ext.lower() not in image_extensions:
        continue

    path = os.path.join(ref_dir, filename)

    # Read reference image exactly like original code
    img = cv2.imread(path)

    if img is None:
        print(f"WARNING: Could not read {filename}")
        continue

    # Resize to 256x256, matching output_size in TrainDataset
    img = cv2.resize(img, output_size)

    # Convert uint8 BGR [0,255] -> float32 BGR [0,1]
    img = img.astype(np.float32) / 255.0

    # Convert to Lab exactly like original code
    lab = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)

    # Calculate mean and std exactly like trans_np()
    mean = np.mean(lab, axis=(0, 1))
    std = np.std(lab, axis=(0, 1))

    # IMPORTANT:
    # Network expects:
    # [L_mean, A_mean, B_mean, L_std, A_std, B_std]
    true_num = np.concatenate([mean, std])

    txt_path = os.path.join(txt_dir, name + ".txt")

    with open(txt_path, "w") as f:
        f.write(",".join(f"{x:.10f}" for x in true_num))

    print(f"{filename}")
    print(f"  mean: {mean}")
    print(f"  std : {std}")
    print(f"  TXT : {txt_path}")

print("\nDone.")
```
